```python
# =====================================================
# TELECOM CUSTOMER CHURN - DATA PREPROCESSING
# =====================================================

# -----------------------------
# 1. Import Libraries
# -----------------------------
import pandas as pd
import numpy as np
import joblib
import warnings

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

# -----------------------------
# 2. Load Dataset
# -----------------------------
df = pd.read_excel(
    r"C:\Customer Churn\Telecom_Churn_Analytics\data\Telco_customer_churn.xlsx"
)

# -----------------------------
# 3. Dataset Overview
# -----------------------------
print("="*60)
print("DATASET SHAPE")
print("="*60)
print(df.shape)

print("\n")

print("="*60)
print("FIRST 5 ROWS")
print("="*60)
print(df.head())

print("\n")

print("="*60)
print("COLUMN NAMES")
print("="*60)
print(df.columns.tolist())

# -----------------------------
# 4. Create Copy
# -----------------------------
df_clean = df.copy()

# -----------------------------
# 5. Missing Values
# -----------------------------
print("\n")
print("="*60)
print("MISSING VALUES")
print("="*60)

print(df_clean.isnull().sum())

# -----------------------------
# 6. Convert Total Charges
# -----------------------------
df_clean["Total Charges"] = pd.to_numeric(
    df_clean["Total Charges"],
    errors="coerce"
)

# -----------------------------
# 7. Fill Missing Values
# -----------------------------
df_clean["Total Charges"].fillna(
    df_clean["Total Charges"].median(),
    inplace=True
)

if "Churn Reason" in df_clean.columns:
    df_clean["Churn Reason"].fillna(
        "No Churn",
        inplace=True
    )

print("\nRemaining Missing Values")
print(df_clean.isnull().sum())

# -----------------------------
# 8. Remove Unnecessary Columns
# -----------------------------
columns_to_drop = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "Lat Long",
    "Latitude",
    "Longitude",
    "Zip Code"
]

df_clean.drop(
    columns=columns_to_drop,
    inplace=True,
    errors="ignore"
)

# -----------------------------
# 9. Convert Target Variable
# -----------------------------
df_clean["Churn Label"] = df_clean["Churn Label"].map({
    "No":0,
    "Yes":1
})

print("\n")
print("="*60)
print("TARGET VARIABLE")
print("="*60)

print(df_clean["Churn Label"].value_counts())

# -----------------------------
# 10. Remove Leakage Columns
# -----------------------------
leakage_columns = [
    "Churn Value",
    "Churn Score",
    "Churn Reason"
]

df_clean.drop(
    columns=leakage_columns,
    inplace=True,
    errors="ignore"
)

# -----------------------------
# 11. Separate Features & Target
# -----------------------------
X = df_clean.drop("Churn Label", axis=1)

y = df_clean["Churn Label"]

# -----------------------------
# 12. One Hot Encoding
# -----------------------------
X = pd.get_dummies(
    X,
    drop_first=True
)

print("\n")
print("="*60)
print("FEATURE MATRIX SHAPE")
print("="*60)
print(X.shape)

# -----------------------------
# 13. Train Test Split
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("\n")
print("="*60)
print("TRAIN TEST SHAPES")
print("="*60)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)

# -----------------------------
# 14. Scale Numerical Features
# -----------------------------
numeric_columns = [
    "Tenure Months",
    "Monthly Charges",
    "Total Charges",
    "CLTV"
]

scaler = StandardScaler()

X_train[numeric_columns] = scaler.fit_transform(
    X_train[numeric_columns]
)

X_test[numeric_columns] = scaler.transform(
    X_test[numeric_columns]
)

# -----------------------------
# 15. Save Files
# -----------------------------
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)

df_clean.to_csv(
    "../data/cleaned_data.csv",
    index=False
)

# -----------------------------
# 16. Final Check
# -----------------------------
print("\n")
print("="*60)
print("PREPROCESSING COMPLETED SUCCESSFULLY")
print("="*60)

print("Final Dataset Shape :", df_clean.shape)
print("Training Samples    :", len(X_train))
print("Testing Samples     :", len(X_test))

print("\n")

print("="*60)
print("FIRST 5 ROWS OF TRAINING DATA")
print("="*60)

print(X_train.head())
```
